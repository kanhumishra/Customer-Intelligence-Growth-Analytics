```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("D:\Customer-Intelligence-Growth-Analytics/data/processed/telco_customer_clean.csv")
```

```python
total_customers = df["CustomerID"].nunique()

print("Total Customers :", total_customers)
```

```python
active_customers = df[df["Churn Label"]=="No"]["CustomerID"].nunique()

print("Active Customers :", active_customers)
```

```python
churned_customers = df[df["Churn Label"]=="Yes"]["CustomerID"].nunique()

print("Churned Customers :", churned_customers)
```

```python
churn_rate = round(
    (churned_customers/total_customers)*100,
    2
)

print("Churn Rate :", churn_rate,"%")
```

```python
total_revenue = round(
    df["Total Charges"].sum(),
    2
)
monthly_revenue = round(
    df["Monthly Charges"].sum(),
    2
)


print("Total Revenue :", total_revenue)
print("Monthly Revenue :", monthly_revenue)
```

```python
avg_monthly_charge = round(
    df["Monthly Charges"].mean(),
    2
)

print(avg_monthly_charge)
```

```python
arpu = round(
    df["Monthly Charges"].mean(),
    2
)

print("Average revenue per user :", arpu)
```

```python
avg_tenure = round(
    df["Tenure Months"].mean(),
    2
)
max_tenure = df["Tenure Months"].max()
min_tenure = df["Tenure Months"].min()

print(max_tenure)
print(min_tenure)
print(avg_tenure)
```

```python
def calculate_kpis(df):
    return {
        "Total Customers": df["CustomerID"].nunique(),
        "Active Customers": df[df["Churn Label"] == "No"]["CustomerID"].nunique(),
        "Churned Customers": df[df["Churn Label"] == "Yes"]["CustomerID"].nunique(),
        "Churn Rate (%)": round((df["Churn Label"] == "Yes").mean() * 100, 2),
        "Total Revenue": round(df["Total Charges"].sum(), 2),
        "Monthly Revenue": round(df["Monthly Charges"].sum(), 2),
        "ARPU": round(df["Monthly Charges"].mean(), 2),
        "Average Tenure": round(df["Tenure Months"].mean(), 2),
        "Average CLTV": round(df["CLTV"].mean(), 2),
        "Average Churn Score": round(df["Churn Score"].mean(), 2)
    }

kpi_summary = pd.DataFrame(
    calculate_kpis(df).items(),
    columns=["KPI", "Value"]
)

kpi_summary
```
