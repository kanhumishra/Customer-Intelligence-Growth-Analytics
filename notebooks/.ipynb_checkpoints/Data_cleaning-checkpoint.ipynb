```python
import numpy as np 
import pandas as pd 
```

```python
df = pd.read_excel("D:\Customer-Intelligence-Growth-Analytics\data\Raw\Telco_customer_churn.xlsx")
```

```python
df_clean = df.copy()
print("Rows and Columns :", df_clean.shape)

df_clean.info()
```

```python
df_clean.isnull().sum()
df_clean.duplicated().sum()
df_clean.drop_duplicates(inplace=True)
```

```python
df.dtypes
```

**Problem is data type of total charge, Have to change to numeric**

```python
df_clean["Total Charges"] = pd.to_numeric(
    df_clean["Total Charges"],
    errors="coerce"
)
df_clean["Total Charges"].dtype
```

```python
df_clean.isnull().sum()
```

```python
df_clean[
    df_clean["Total Charges"].isna()
][["CustomerID", "Tenure Months", "Monthly Charges", "Total Charges"]]
```

```python
df_clean["Total Charges"] = df_clean["Total Charges"].fillna(0)
```

```python
for col in df_clean.select_dtypes(include="object"):
    print(f"\n{col}")
    print(df_clean[col].unique())
```

```python
object_cols = df_clean.select_dtypes(include="object").columns

for col in object_cols:
    df_clean[col] = df_clean[col].str.strip()
```

```python
df_clean.info()

df_clean.describe(include="all")
```

```python
df_clean.to_csv(
    "D:\Customer-Intelligence-Growth-Analytics/data/processed/telco_customer_clean.csv",
    index=False
)
```
