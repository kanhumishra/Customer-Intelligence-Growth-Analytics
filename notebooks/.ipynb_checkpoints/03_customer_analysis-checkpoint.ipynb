```python
import pandas as pd
df = pd.read_csv("D:\Customer-Intelligence-Growth-Analytics/data/processed/telco_customer_clean.csv")
df["CustomerID"].nunique()
```

```python
gender = (
    df["Gender"]
      .value_counts()
      .reset_index()
)

gender.columns = ["Gender","Customers"]

gender
```

***Here we can see the total customers having appox same value from both gender***

```python
churn_gender = (
    df[df["Churn Label"] == "Yes"]["Gender"]
      .value_counts()
      .reset_index()
)
churned_customers = df[df["Churn Label"]=="Yes"]["CustomerID"].nunique()


churn_gender.columns = ["Gender", "Churned Customers"]
print("Churned Customers :", churned_customers)
churn_gender
```

```python
#Here we can see that there is no gender related issue 
import matplotlib.pyplot as plt

gender.plot(
    x="Gender",
    y="Customers",
    kind="bar"
)

plt.title("Customer Distribution by Gender")

plt.show()
```

```python
senior = (
    df["Senior Citizen"]
      .value_counts()
      .reset_index()
)

senior.columns = ["Senior Citizen","Customers"]
senior["Senior Citizen"] = senior["Senior Citizen"].replace({
    0:"No",
    1:"Yes"
})
senior
#here we can see very less senior citizen are our customer 
```
